## Ejemplo — Reconocer dígitos (0 a 9) con regresión logística
Usamos el dataset de dígitos pequeño de scikit-learn (imágenes 8×8).

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt

digits = load_digits()
plt.gray()
plt.matshow(digits.images[0]); plt.title('Primer dígito: %d' % digits.target[0]); plt.show()
```

```python
X_train,X_test,y_train,y_test = train_test_split(digits.data, digits.target, test_size=0.2, random_state=2)
model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)
print('Precisión (acierto en 10 clases):', round(model.score(X_test, y_test),4))
```

```python
import numpy as np
# Predecir el primer dígito de prueba
pred = model.predict([X_test[0]])[0]
print('El modelo predice:', pred, ' | Real:', y_test[0])
```

**Resultado:** para varias categorías, el modelo calcula la probabilidad de cada clase (0–9) y elige la más alta. Es el paso previo a las redes neuronales.
