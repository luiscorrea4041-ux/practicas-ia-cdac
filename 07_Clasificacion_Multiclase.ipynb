Ahora hay **más de dos** clases. Clasificaremos imágenes de **dígitos escritos a mano** (0 a 9) con regresión logística, usando el dataset `digits` de scikit-learn.

## 1. Cargamos las imágenes de dígitos

Cada imagen es de 8×8 píxeles (64 números) y representa un dígito del 0 al 9.

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
```

Mostramos las primeras 5 imágenes para ver cómo lucen los dígitos.

```python
plt.gray()
for i in range(5):
    plt.matshow(digits.images[i])
    plt.show()
```

Vemos qué contiene el dataset y cómo se ven los datos de una imagen (64 valores de intensidad).

```python
dir(digits)
```

```python
digits.data[0]
```

## 2. Creamos y entrenamos el modelo

Separamos 80% entrenamiento y 20% prueba, y entrenamos la regresión logística.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2, random_state=42)
len(X_train)
```

```python
model = LogisticRegression(max_iter=5000)
model.fit(X_train, y_train)
```

## 3. Medimos la precisión

Qué porcentaje de dígitos clasifica bien en el conjunto de prueba.

```python
model.score(X_test, y_test)
```

Probamos el modelo prediciendo los primeros 5 dígitos del dataset.

```python
model.predict(digits.data[0:5])
```

## 4. Matriz de confusión

La **matriz de confusión** muestra, para cada dígito real, cuántos fueron clasificados en cada clase. La diagonal son los aciertos.

```python
y_predicted = model.predict(X_test)
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm, annot=True)
plt.xlabel('Predicho')
plt.ylabel('Real')
plt.show()
```

## Ejercicio práctico

**Enunciado (del curso):** Usar el dataset **iris** de `sklearn.datasets` para entrenar un modelo de regresión logística. Hay 150 flores con 4 características (largo/ancho de sépalo y pétalo) y 3 clases: **Setosa, Versicolour, Virginica**. Calcular la precisión y la matriz de confusión.

**Paso 1.** Cargamos el dataset iris.

```python
from sklearn.datasets import load_iris
iris = load_iris()
print("Características:", iris.feature_names)
print("Clases:", list(iris.target_names))
```

**Paso 2.** Dividimos en entrenamiento y prueba, y entrenamos el modelo.

```python
Xtr, Xte, ytr, yte = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
model_iris = LogisticRegression(max_iter=200)
model_iris.fit(Xtr, ytr)
```

**Paso 3.** Medimos la precisión.

```python
print(f"Precisión en iris: {model_iris.score(Xte, yte):.4f}")
```

**Paso 4.** Matriz de confusión del clasificador de flores.

```python
cm_iris = confusion_matrix(yte, model_iris.predict(Xte))
plt.figure(figsize=(6,5))
sn.heatmap(cm_iris, annot=True, xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.xlabel('Predicho'); plt.ylabel('Real')
plt.title('Matriz de confusión - Iris')
plt.show()
```

**Conclusión:** la regresión logística clasifica las flores iris con una precisión muy alta (cercana al 100%). La matriz de confusión muestra que casi todas las flores caen en la diagonal (aciertos); la clase *setosa* se separa perfectamente y la pequeña confusión ocurre entre *versicolor* y *virginica*, que son más parecidas.
